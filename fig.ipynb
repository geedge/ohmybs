```python
# 导入包
import plotly.express as px
import numpy as np
import pandas as pd
import datetime
import random
import plotly.offline as py  #设置离线画图
import plotly.io as pio
```

```python
# 水印效果比较
# 日期数据
begin = datetime.date(2023,12,6)
end = datetime.date(2023,12,20)
datelist=[]
mydata=[]
olddata=[]
for i in range((end - begin).days+1):
    day = begin + datetime.timedelta(days=i)
    datelist.append(day.strftime("%Y年%m月%d日"))
    mydata.append([200+int(i*random.choice([204,208,201,206]))+random.randint(-50, 50),"动态水印（本方法）---  "])
    olddata.append([100+int(i*random.choice([107,103,106,102]))+random.randint(-20,20),"静态水印          "])
data=[]
for i in range(len(datelist)):
    data.append([datelist[i],mydata[i][0],mydata[i][1]])
    data.append([datelist[i],olddata[i][0],olddata[i][1]])
```

```python
# 数据
df=pd.DataFrame(data,columns=["日期","收集IPv6地址数量","水印类型"])
```

```python
# 绘图
fig = px.line(df, x="日期", y="收集IPv6地址数量",markers=True,symbol="水印类型",color="水印类型",template="plotly_white")
fig.update_yaxes(
    zeroline=True,
    linecolor='black',
    showline=True,
    automargin=True
)
fig.update_xaxes(
    
    linecolor='black',
    showline=True,
    automargin=True
)
fig.update_layout(
        legend=dict(
        x=1,
        y=1,
        title_font_family="Times New Roman",
        bordercolor="Black",
        borderwidth=2
    )
)
fig.write_html("images/水印比较.html")
fig
```
